```python
import matplotlib.pyplot as plt
import numpy as np

# Create data points
x = np.linspace(0, 5, 200)

# Define the functions
y_linear = x
y_quadratic = x**2
y_cubic = x**3

# Create subplots: 3 rows, 1 column
fig, axs = plt.subplots(1, 3, figsize=(15,5))

# Plot Linear function in the first subplot
axs[0].plot(x, y_linear, label='Linear', color='blue')
axs[0].set_title('Linear Function')
axs[0].set_xlabel('x')
axs[0].set_ylabel('y')
axs[0].set_xticks([0, 2, 4])

# Plot Quadratic function in the second subplot
axs[1].plot(x, y_quadratic, label='Quadratic', color='green')
axs[1].set_title('Quadratic Function')
axs[1].set_xlabel('x')
axs[1].set_ylabel('y')
axs[1].set_xticks([0, 2, 4])

# Plot Cubic function in the third subplot
axs[2].plot(x, y_cubic, label='Cubic', color='red')
axs[2].set_title('Cubic Function')
axs[2].set_xlabel('x')
axs[2].set_ylabel('y')
axs[2].set_xticks([0, 2, 4])

# Add space between subplots
plt.tight_layout()

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
a, b = 0, 1
data = np.random.normal(a, b, 1000)
plt.figure(figsize=(8, 6))
count, bins, ignored = plt.hist(data, bins=30, density=True, alpha=0.6, color='green', label='Histogram')
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, a, b)
plt.plot(x, p, 'k', linewidth=2, label='PDF (Normal Distribution)')
plt.title('Histogram of Random Numbers with Normal PDF Overlay')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
import numpy as np

# Define the matrix A
A = np.array([[4, -2], 
              [1, 1]])  # A 2x2 matrix

# Display the matrix
print("Matrix A:")
print(A)

# Calculate the determinant of A
det_A = np.linalg.det(A)
print("\nDeterminant of A:", det_A)

# Calculate the inverse of A
try:
    inv_A = np.linalg.inv(A)
    print("\nInverse of A:")
    print(inv_A)
except np.linalg.LinAlgError:
    print("\nMatrix A is singular and cannot be inverted.")

# Transpose the matrix A
transpose_A = A.T
print("\nTranspose of A:")
print(transpose_A)

# Perform matrix multiplication with another matrix B
B = np.array([[2], 
              [3]])  # A 2x1 matrix
C = A @ B  # Matrix multiplication
print("\nMatrix C (A multiplied by B):")
print(C)
```

```python
import numpy as np

# Define the coefficient matrix A
A = np.array([[2, -1, 3],
              [1, -3, -2],
              [3, 2, -1]])

# Define the constant vector b
b = np.array([9, 0, -1])

# Solve the system of equations
solution = np.linalg.solve(A, b)

# Display the solution
print("Solution (x, y, z):")
print(solution)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n_flips = 10         # Number of coin flips
p_heads = 0.5        # Probability of getting heads
n_experiments = 1000 # Number of experiments

# Simulate coin flips
# Generate random numbers and determine number of heads in each experiment
results = np.random.binomial(n_flips, p_heads, n_experiments)

# Count occurrences of heads (0 to 10)
counts = np.bincount(results)

# Calculate PMF
x = np.arange(len(counts)) # Possible number of heads
pmf = counts / n_experiments # Normalize counts to get PMF

# Plotting the PMF
plt.figure(figsize=(10, 6))
plt.bar(x, pmf, color='blue', alpha=0.7, label='Simulated PMF')
plt.title('Probability Mass Function of Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of Heads')
plt.ylabel('Probability')
plt.xticks(x)
plt.grid(axis='y')

# Overlay the theoretical PMF for comparison
x_theory = np.arange(0, n_flips + 1)
pmf_theory = binom.pmf(x_theory, n_flips, p_heads)
plt.plot(x_theory, pmf_theory, 'ro-', label='Theoretical PMF', markersize=8)

# Add legend
plt.legend()
plt.show()
```

```python
import pandas as pd
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C'],
    'Sales': [100, 150, 200, 120, 130, 170],
    'Month': ['Jan', 'Jan', 'Jan', 'Feb', 'Feb', 'Feb']
}

df = pd.DataFrame(data)
sales = df.groupby('Product')['Sales'].sum().reset_index()
sales.columns = ['Product', 'Total Sales']

# Display total sales for each product
print("Total Sales for Each Product:")
print(sales)

# Step 3: Create a pivot table showing total sales for each product per month
pivot_table = df.pivot_table(values='Sales', index='Month', columns='Product', aggfunc='sum', fill_value=0)

# Display the pivot table
print("\nPivot Table of Total Sales per Product per Month:")
print(pivot_table)

# Step 4: Calculate the percentage of total sales for each row in the original DataFrame
total_sales = df['Sales'].sum()
df['Percentage of Total Sales'] = (df['Sales'] / total_sales) * 100

# Display the updated DataFrame
print("\nUpdated DataFrame with Percentage of Total Sales:")
print(df)
```
