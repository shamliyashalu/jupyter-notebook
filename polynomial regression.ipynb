```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
np.random.seed(42)b
x = np.random.rand(100, 1) * 10 
y = 2 + 3 * x**2 + np.random.randn(100, 1) * 10 
```

```python
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x)
model = LinearRegression()
model.fit(x_poly, y)
```

```python
y_pred = model.predict(x_poly)
```

```python
plt.scatter(x, y, color='blue', label='Original Data')
plt.plot(x, y_pred, color='red', label='Polynomial Fit')
plt.title('Polynomial Regression Degree 2')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
mse = mean_squared_error(y, y_pred)
print(f'Mean Squared Error: {mse:.2f}')
```

```python
r2 = r2_score(y, y_pred)
print(f'R-squared: {r2:.2f}')
```

## Given a dataset that shows the relationship between the size of a house (in square feet) and 
## its price, use polynomial regression to predict housing prices based on the size.

```python
np.random.seed(42)
size = np.random.rand(100, 1) * 2000  
price = 50000 + 10 * size + 0.02 * size**2 + np.random.randn(100, 1) * 20000  
plt.scatter(size, price, color='blue')
plt.title('House Size vs Price')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price (USD)')
plt.show()
```

```python
poly = PolynomialFeatures(degree=2)
size_poly = poly.fit_transform(size)
model = LinearRegression()
model.fit(size_poly, price)
price_pred = model.predict(size_poly)
plt.scatter(size, price, color='blue', label='Original Data')
plt.plot(size, price_pred, color='red', label='Polynomial Fit (Degree 2)')
plt.title('Polynomial Regression Fit (Degree 2) - Housing Prices')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()
```

```python
poly = PolynomialFeatures(degree=1)
size_poly = poly.fit_transform(size)
model = LinearRegression()
model.fit(size_poly, price)
price_pred = model.predict(size_poly)
plt.scatter(size, price, color='blue', label='Original Data')
plt.plot(size, price_pred, color='red', label='Polynomial Fit (Degree 2)')
plt.title('Polynomial Regression Fit (Degree 2) - Housing Prices')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()
```

```python
poly = PolynomialFeatures(degree=3)
size_poly = poly.fit_transform(size)
model = LinearRegression()
model.fit(size_poly, price)
price_pred = model.predict(size_poly)
plt.scatter(size, price, color='blue', label='Original Data')
plt.plot(size, price_pred, color='red', label='Polynomial Fit (Degree 2)')
plt.title('Polynomial Regression Fit (Degree 2) - Housing Prices')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()
```

```python
mse = mean_squared_error(price, price_pred)
print(f'Mean Squared Error: {mse:.2f}')
```

```python
r2 = r2_score(price, price_pred)
print(f'R-squared: {r2:.2f}')
```

# 2 Suppose you are given the population data for a city over 
# several decades, and you want to model population growth using polynomial regression. 
# The growth is not linear, so polynomial regression is used to capture the trend.

```python
np.random.seed(42)
years = np.arange(1960, 2020).reshape(-1, 1)  #  from 1960 to 2020
population = 20000 + 500 * (years - 1960) + 2 * (years - 1960)**2 + np.random.randn(len(years), 1) * 10000  
plt.scatter(years, population, color='green')
plt.title('Year vs Population Growth')
plt.xlabel('Year')
plt.ylabel('Population')
plt.show()
```

```python
poly = PolynomialFeatures(degree=2)
years_poly = poly.fit_transform(years)
model = LinearRegression()
model.fit(years_poly, population)
population_pred = model.predict(years_poly)
plt.scatter(years, population, color='green', label='Original Data')
plt.plot(years, population_pred, color='red', label='Polynomial Fit (Degree 2)')
plt.title('Polynomial Regression Fit (Degree 2) - Population Growth')
plt.xlabel('Year')
plt.ylabel('Population')
plt.legend()
plt.show()
```

```python
mse = mean_squared_error(population, population_pred)
r2 = r2_score(population, population_pred)
print(f'Mean Squared Error: {mse:.2f}')
print(f'R-squared: {r2:.2f}')
```

#  Suppose you have data showing stock prices over time, and the trend appears 
# to be more complex than linear or quadratic.
# Fit a higher-degree polynomial regression model to better capture the fluctuations.

```python
np.random.seed(42)
days = np.linspace(1, 100, 100).reshape(-1, 1)  #  from 1 to 100
price = 100 + 0.5 * days + 0.05 * days**2 - 0.0003 * days**3 + np.random.randn(len(days), 1) * 5 
plt.scatter(days, price, color='purple')
plt.title('Days vs Stock Price')
plt.xlabel('Days')
plt.ylabel('Stock Price')
plt.show()
```

```python
poly = PolynomialFeatures(degree=3)
days_poly = poly.fit_transform(days)
model = LinearRegression()
model.fit(days_poly, price)
price_pred = model.predict(days_poly)
plt.scatter(days, price, color='purple', label='Original Data')
plt.plot(days, price_pred, color='orange', label='Polynomial Fit (Degree 3)')
plt.title('Polynomial Regression Fit (Degree 3) - Stock Prices')
plt.xlabel('Days')
plt.ylabel('Stock Price')
plt.legend()
plt.show()
```

```python
mse = mean_squared_error(price, price_pred)
r2 = r2_score(price, price_pred)

print(f'Mean Squared Error: {mse:.2f}')
print(f'R-squared: {r2:.2f}')
```

# Investigate the risk of overfitting when using a very high-degree polynomial (e.g., degree 10) on simple data.

```python
# Generate simple data (linear trend)
np.random.seed(42)
x_simple = np.linspace(1, 10, 100).reshape(-1, 1)  # X values from 1 to 10
y_simple = 2 * x_simple + np.random.randn(100, 1) * 2  # Linear relationship with noise

# Scatter plot of the data
plt.scatter(x_simple, y_simple, color='blue')
plt.title('Simple Data (Linear Relationship)')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
poly = PolynomialFeatures(degree=10)
x_poly = poly.fit_transform(x_simple)
model = LinearRegression()
model.fit(x_poly, y_simple)
y_pred = model.predict(x_poly)
plt.scatter(x_simple, y_simple, color='blue', label='Original Data')
plt.plot(x_simple, y_pred, color='red', label='Polynomial Fit (Degree 10)')
plt.title('Overfitting with High-Degree Polynomial (Degree 10)')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
mse = mean_squared_error(y_simple, y_pred)
r2 = r2_score(y_simple, y_pred)

print(f'Mean Squared Error: {mse:.2f}')
print(f'R-squared: {r2:.2f}')
```

```python
np.random.seed(0)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)
residuals = y - y_pred

plt.scatter(X, residuals, color='blue', label='Residuals')
plt.axhline(y=0, color='red', linestyle='--', label='Zero Line')
plt.title('Residual Plot')
plt.xlabel('X')
plt.ylabel('Residuals')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error, r2_score

# Calculate R-squared and Adjusted R-squared
r2 = r2_score(y, y_pred)
n = len(y)  # Number of samples
p = X.shape[1]  # Number of predictors (for Adjusted R-squared)
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)

# Calculate MSE and RMSE
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

print(f"R-squared: {r2:.4f}")
print(f"Adjusted R-squared: {adjusted_r2:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
```
