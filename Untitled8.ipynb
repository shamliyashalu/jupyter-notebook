```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
sales_data = [150, 200, 250, 210, 270, 300, 320, 310, 290, 280, 270, 260]
months = pd.date_range(start='2021-01-01', periods=len(sales_data), freq='M')
sales_series = pd.Series(sales_data, index=months)

# Fit an ARIMA model
model = ARIMA(sales_series, order=(1, 1, 1))  # Adjust the order as needed
model_fit = model.fit()
```
